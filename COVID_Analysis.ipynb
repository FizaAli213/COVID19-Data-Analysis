```python
clean_df = df.fillna(0)
```

```python
highest_cases = clean_df.orderBy(
    clean_df.total_cases.desc()
)

highest_cases.select(
    "country",
    "total_cases"
).show(10)
```

```python
highest_cases = clean_df.filter(
    clean_df.country != "World"
).orderBy(
    clean_df.total_cases.desc()
)

highest_cases.select(
    "country",
    "total_cases"
).show(10)
```

```python
highest_cases = clean_df.filter(
    (~clean_df.country.contains("World")) &
    (~clean_df.country.contains("income")) &
    (~clean_df.country.contains("Europe")) &
    (~clean_df.country.contains("Asia")) &
    (~clean_df.country.contains("Africa")) &
    (~clean_df.country.contains("Oceania"))
).orderBy(
    clean_df.total_cases.desc()
)

highest_cases.select(
    "country",
    "total_cases"
).show(10)
```

```python
highest_cases = clean_df.filter(
    (~clean_df.country.contains("World")) &
    (~clean_df.country.contains("income")) &
    (clean_df.country != "North America") &
    (clean_df.country != "South America") &
    (clean_df.country != "Europe") &
    (clean_df.country != "Asia") &
    (clean_df.country != "Africa") &
    (clean_df.country != "Oceania") &
    (clean_df.country != "European Union")
).orderBy(
    clean_df.total_cases.desc()
)

highest_cases.select(
    "country",
    "total_cases"
).show(10)
```

```python
highest_cases = clean_df.filter(
    clean_df.continent.isNotNull()
).orderBy(
    clean_df.total_cases.desc()
)

highest_cases.select(
    "country",
    "total_cases"
).show(10)
```

```python
from pyspark.sql.functions import max

highest_cases = clean_df.groupBy("country").agg(
    max("total_cases").alias("highest_cases")
)

highest_cases.orderBy(
    highest_cases.highest_cases.desc()
).show(10)
```
